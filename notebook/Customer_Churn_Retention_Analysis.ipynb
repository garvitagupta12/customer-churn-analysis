```python
categorical_columns = [
    "Geography",
    "Gender",
    "NumOfProducts",
    "HasCrCard",
    "IsActiveMember",
    "Exited"
]

for col in categorical_columns:
    print(f"\n--- {col} ---")
    print(df[col].value_counts())
```

```python
geography_churn = (
    df.groupby("Geography")["Exited"]
    .agg(["count", "sum", "mean"])
    .reset_index()
)

geography_churn.columns = [
    "Geography",
    "Total_Customers",
    "Churned_Customers",
    "Churn_Rate"
]

geography_churn["Churn_Rate"] *= 100

geography_churn.sort_values(
    "Churn_Rate",
    ascending=False
)
```

### Key Finding — Geography

Germany has the highest customer churn rate at 32.44%, which is approximately
twice the churn rate observed in France (16.15%) and Spain (16.67%).

```python
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[17, 25, 35, 45, 55, 65, 100],
    labels=[
        "18-25",
        "26-35",
        "36-45",
        "46-55",
        "56-65",
        "66+"
    ]
)

age_churn = (
    df.groupby("Age_Group", observed=False)["Exited"]
    .agg(["count", "sum", "mean"])
    .reset_index()
)

age_churn.columns = [
    "Age_Group",
    "Total_Customers",
    "Churned_Customers",
    "Churn_Rate"
]

age_churn["Churn_Rate"] *= 100

age_churn
```

##Key Finding — Age
Customer churn varies substantially across age groups.

Customers aged 46-55 have the highest churn rate at 50.57%, followed by customers aged 56-65 at 48.32%. In comparison, younger customers aged
18-25 and 26-35 have much lower churn rates of 7.53% and 8.50%.

```python
gender_churn = (
    df.groupby("Gender")["Exited"]
    .agg(["count", "sum", "mean"])
    .reset_index()
)

gender_churn.columns = [
    "Gender",
    "Total_Customers",
    "Churned_Customers",
    "Churn_Rate"
]

gender_churn["Churn_Rate"] *= 100

gender_churn.sort_values("Churn_Rate", ascending=False)
```

## Key Finding — Gender

Female customers have a higher observed churn rate (25.07%) than male
customers (16.46%), representing an 8.62 percentage-point difference.

```python
activity_churn = (
    df.groupby("IsActiveMember")["Exited"]
    .agg(["count", "sum", "mean"])
    .reset_index()
)

activity_churn.columns = [
    "IsActiveMember",
    "Total_Customers",
    "Churned_Customers",
    "Churn_Rate"
]

activity_churn["Churn_Rate"] *= 100

activity_churn
```

## Key Finding — Customer Activity

Inactive customers have a substantially higher observed churn rate (26.85%)
than active customers (14.27%), representing a 12.58 percentage-point
difference.

```python
credit_card_churn = (
    df.groupby("HasCrCard")["Exited"]
    .agg(["count", "sum", "mean"])
    .reset_index()
)

credit_card_churn.columns = [
    "HasCrCard",
    "Total_Customers",
    "Churned_Customers",
    "Churn_Rate"
]

credit_card_churn["Churn_Rate"] *= 100

credit_card_churn
```

## Key Finding — Credit Card Ownership

Churn rates are very similar between customers with and without a credit
card. Customers without a credit card have a churn rate of 20.81%, compared
with 20.18% among customers with a credit card.

```python
tenure_churn = (
    df.groupby("Tenure")["Exited"]
    .agg(["count", "sum", "mean"])
    .reset_index()
)

tenure_churn.columns = [
    "Tenure",
    "Total_Customers",
    "Churned_Customers",
    "Churn_Rate"
]

tenure_churn["Churn_Rate"] *= 100

tenure_churn
```

## Key Finding — Customer Tenure

Churn rates remain relatively stable across different customer tenure
levels, ranging from 17.22% to 23.00%

```python
df["CreditScore_Group"] = pd.cut(
    df["CreditScore"],
    bins=[300, 580, 670, 740, 851],
    labels=["Poor", "Fair", "Good", "Excellent"]
)

credit_churn = (
    df.groupby("CreditScore_Group", observed=False)["Exited"]
    .agg(["count", "sum", "mean"])
    .reset_index()
)

credit_churn.columns = [
    "CreditScore_Group",
    "Total_Customers",
    "Churned_Customers",
    "Churn_Rate"
]

credit_churn["Churn_Rate"] *= 100

credit_churn
```

## Key Finding — Credit Score

Churn rates are relatively similar across credit score groups, ranging from
18.82% for customers with Good credit scores to 22.15% for customers with
Poor credit scores.

## Key EDA Insights

The exploratory analysis identified several important patterns associated
with customer churn:

1. **Age is a major churn-related factor.**
   Customers aged 46-55 have the highest churn rate at 50.57%, followed by
   the 56-65 group at 48.32%.

2. **Customer activity is strongly associated with retention.**
   Inactive customers have a churn rate of 26.85%, compared with 14.27% for
   active customers.

3. **Geography shows a significant difference in churn.**
   Germany has the highest churn rate at 32.44%, approximately twice the
   churn rate observed in France and Spain.

4. **Gender shows a noticeable difference.**
   Female customers have a churn rate of 25.07%, compared with 16.46% for
   male customers.

5. **Credit score appears to have a relatively weak relationship with churn.**
   Churn rates across credit-score groups remain relatively close.

6. **Customer tenure does not show a clear trend.**
   Churn rates fluctuate between 17.22% and 23.00% across tenure levels.

7. **Credit-card ownership has minimal association with churn.**
   Churn rates differ by only 0.63 percentage points between customers with
   and without a credit card.

These findings suggest that retention efforts should primarily investigate
customer age, engagement/activity, and geography, while treating other
variables as supporting factors rather than standalone churn indicators.
